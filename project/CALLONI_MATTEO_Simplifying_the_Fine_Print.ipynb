```python


import os
import sys

# INSTALLAZIONE LIBRERIE
print(" Installazione delle librerie necessarie")
!pip install -q transformers accelerate bitsandbytes
!pip install -q google-generativeai
!pip install -q sentence-transformers bert-score
!pip install -q textstat datasets
!pip install -q matplotlib seaborn pandas numpy scipy tqdm

import torch
import gc
import time
import textstat
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from tqdm.notebook import tqdm
from scipy import stats
from datasets import load_dataset
import google.generativeai as genai
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig, pipeline
from sentence_transformers import SentenceTransformer, util
from bert_score import score as bert_score_func
from google.colab import userdata, output

# Gestione Warnings
import warnings
warnings.filterwarnings('ignore')

# Configurazione Display Pandas
pd.set_option('display.max_colwidth', None)

# CLASSE PRINCIPALE

class LegalSimplifier:
    def __init__(self, google_api_key, hf_token):
        """
        Inizializza il semplificatore con le chiavi API necessarie.
        """
        self.google_api_key = google_api_key
        self.hf_token = hf_token
        self.dataset = None
        self.results = []

        # Configurazione Gemini
        genai.configure(api_key=self.google_api_key)

        # Prompt Fisso richiesto
        self.system_prompt_template = """
You are a legal expert translator. Simplify the following legal text for a general non-expert audience (8th-grade reading level) while strictly preserving all legal obligations, rights, limitations, and conditions. Do not omit any critical information.

Legal Text:
{text}

Simplified Version:
"""

    def load_dataset(self, num_samples=100):
        """
        Carica e preprocessa il dataset (CUAD o LexGLUE come fallback).
        Filtra paragrafi tra 50 e 300 parole.
        """
        print(f"\n Caricamento dataset...")
        try:
            data = load_dataset("cuad", split="train", streaming=True)
            print("   Dataset CUAD caricato.")
            key_text = "context"
        except Exception:
            # Fallback su LexGLUE se CUAD fallisce o è troppo lento
            print("   CUAD non disponibile, switch su LexGLUE (unfair_tos).")
            data = load_dataset("lex_glue", "unfair_tos", split="train", streaming=True)
            key_text = "text"

        processed_data = []
        iterator = iter(data)

        pbar = tqdm(total=num_samples, desc="Estrazione testi validi")

        seen_texts = set()

        while len(processed_data) < num_samples:
            try:
                item = next(iterator)
                text = item[key_text]

                # Pulizia base
                text = text.replace("\n", " ").strip()
                word_count = len(text.split())

                # Filtro lunghezza e duplicati
                if 50 <= word_count <= 300 and text not in seen_texts:
                    seen_texts.add(text)
                    processed_data.append(text)
                    pbar.update(1)
            except StopIteration:
                break

        self.dataset = processed_data
        print(f" Dataset pronto: {len(self.dataset)} paragrafi caricati.")

    def simplify_with_gemini(self):
        """
        Genera semplificazioni usando Google Gemini 2.5 Flash (o 1.5 Flash se 2.5 non disp).
        Gestisce rate limiting con retry logic.
        """
        print(f"\n Avvio generazione con Gemini...")
        # Usa il modello flash più recente disponibile
        model_name = "gemini-1.5-flash"
        model = genai.GenerativeModel(model_name)
        generation_config = genai.types.GenerationConfig(temperature=0.3)

        gemini_outputs = []

        for text in tqdm(self.dataset, desc="Gemini Processing"):
            prompt = self.system_prompt_template.format(text=text)
            retries = 0
            max_retries = 3
            success = False

            while not success and retries < max_retries:
                try:
                    response = model.generate_content(
                        prompt,
                        generation_config=generation_config
                    )
                    gemini_outputs.append(response.text.strip())
                    success = True
                    time.sleep(1.1) # Rispetto rate limit
                except Exception as e:
                    retries += 1
                    wait_time = 2 ** retries
                    # print(f"Error Gemini: {e}. Retrying in {wait_time}s...")
                    time.sleep(wait_time)

            if not success:
                gemini_outputs.append("ERROR_GENERATION")

        return gemini_outputs

    def simplify_with_llama(self):
        """
        Genera semplificazioni con Llama 3 8B Quantizzato a 4-bit.
        Include gestione memoria aggressiva per Colab Free.
        """
        print(f"\n Avvio generazione con Llama 3 8B (4-bit)...")

        model_id = "meta-llama/Meta-Llama-3-8B-Instruct"

        # Configurazione quantizzazione
        bnb_config = BitsAndBytesConfig(
            load_in_4bit=True,
            bnb_4bit_use_double_quant=True,
            bnb_4bit_quant_type="nf4",
            bnb_4bit_compute_dtype=torch.float16
        )

        try:
            tokenizer = AutoTokenizer.from_pretrained(model_id, token=self.hf_token)
            tokenizer.pad_token = tokenizer.eos_token

            model = AutoModelForCausalLM.from_pretrained(
                model_id,
                quantization_config=bnb_config,
                device_map="auto",
                token=self.hf_token
            )
        except Exception as e:
            print(f" Errore caricamento Llama: {e}")
            print("Assicurati di aver accettato la licenza su HuggingFace e che il token sia corretto.")
            return ["ERROR_LLAMA"] * len(self.dataset)

        llama_outputs = []

        pipe = pipeline(
            "text-generation",
            model=model,
            tokenizer=tokenizer,
            max_new_tokens=512,
            temperature=0.3,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )

        for text in tqdm(self.dataset, desc="Llama Processing"):
            prompt = self.system_prompt_template.format(text=text)

            # Formattazione chat template per Llama 3
            messages = [
                {"role": "system", "content": "You are a helpful legal assistant."},
                {"role": "user", "content": prompt},
            ]
            prompt_formatted = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)

            try:
                outputs = pipe(prompt_formatted)
                generated_text = outputs[0]["generated_text"]
                # Estrai solo la risposta
                response = generated_text.split("assistant<|end_header_id|>\n\n")[-1].strip()
                llama_outputs.append(response)
            except Exception as e:
                llama_outputs.append("ERROR_GENERATION")

        # PULIZIA MEMORIA
        del model
        del tokenizer
        del pipe
        gc.collect()
        torch.cuda.empty_cache()
        print("Memoria GPU liberata dopo Llama.")

        return llama_outputs

    def calculate_metrics(self, original_list, simplified_list, model_name):
        """
        Calcola FKGL, Gunning Fog, BERTScore e Cosine Similarity.
        """
        print(f" Calcolo metriche per {model_name}...")

        # Carica modelli per similarità (solo ora per risparmiare RAM)
        sbert_model = SentenceTransformer('all-MiniLM-L6-v2')

        metrics_data = []

        # Calcolo embedding in batch
        orig_embeddings = sbert_model.encode(original_list, convert_to_tensor=True)
        simp_embeddings = sbert_model.encode(simplified_list, convert_to_tensor=True)
        cosine_scores = util.cos_sim(orig_embeddings, simp_embeddings).diagonal().cpu().numpy()

        # Calcolo BERTScore
        try:
            P, R, F1 = bert_score_func(simplified_list, original_list, lang="en", verbose=False, device='cuda' if torch.cuda.is_available() else 'cpu', batch_size=16)
            bert_f1_scores = F1.numpy()
        except Exception as e:
            print(f"Warning BERTScore: {e}. Using dummy values.")
            bert_f1_scores = [0.0] * len(original_list)

        for i, (orig, simp) in enumerate(zip(original_list, simplified_list)):
            # Readability
            try:
                fkgl_orig = textstat.flesch_kincaid_grade(orig)
                fkgl_simp = textstat.flesch_kincaid_grade(simp)
                delta_fkgl = fkgl_orig - fkgl_simp # Quanto è migliorato

                fog_simp = textstat.gunning_fog(simp)
            except:
                delta_fkgl = 0
                fog_simp = 0

            metrics_data.append({
                'original_fkgl': fkgl_orig,
                'simplified_fkgl': fkgl_simp,
                'delta_fkgl': delta_fkgl,
                'gunning_fog': fog_simp,
                'bert_score_f1': bert_f1_scores[i],
                'cosine_similarity': cosine_scores[i]
            })

        # Pulizia memoria SBERT
        del sbert_model
        gc.collect()
        torch.cuda.empty_cache()

        return metrics_data

    def run_experiment(self):
        """
        Esegue l'intera pipeline: caricamento, inferenza, metriche, salvataggio.
        """
        # Carica Dati
        self.load_dataset(num_samples=100) # O 20 per test rapido

        # Inferenza Llama (prima Llama perché occupa più VRAM)
        llama_preds = self.simplify_with_llama()

        # Inferenza Gemini
        gemini_preds = self.simplify_with_gemini()

        # Calcolo Metriche
        gemini_metrics = self.calculate_metrics(self.dataset, gemini_preds, "Gemini")
        llama_metrics = self.calculate_metrics(self.dataset, llama_preds, "Llama")

        # Costruzione DataFrame Finale
        data_rows = []
        for i in range(len(self.dataset)):
            row = {
                'text_id': i,
                'original_text': self.dataset[i],
                'gemini_simplified': gemini_preds[i],
                'llama_simplified': llama_preds[i],
                # Metriche Gemini
                'gemini_delta_fkgl': gemini_metrics[i]['delta_fkgl'],
                'gemini_bert_f1': gemini_metrics[i]['bert_score_f1'],
                'gemini_cosine': gemini_metrics[i]['cosine_similarity'],
                'gemini_fog': gemini_metrics[i]['gunning_fog'],
                # Metriche Llama
                'llama_delta_fkgl': llama_metrics[i]['delta_fkgl'],
                'llama_bert_f1': llama_metrics[i]['bert_score_f1'],
                'llama_cosine': llama_metrics[i]['cosine_similarity'],
                'llama_fog': llama_metrics[i]['gunning_fog']
            }
            data_rows.append(row)

        self.df = pd.DataFrame(data_rows)

        # Salvataggio CSV
        self.df.to_csv("legal_simplification_results.csv", index=False)
        print("\n Risultati salvati in 'legal_simplification_results.csv'")

        return self.df

    def visualize_results(self):
        """
        Genera i grafici richiesti (Scatter, Boxplot, Heatmap) e analisi statistica.
        """
        if self.df is None:
            print("Nessun risultato da visualizzare.")
            return

        sns.set_style("whitegrid")

        # Scatter Plot
        plt.figure(figsize=(10, 6))

        # Gemini Data
        plt.scatter(self.df['gemini_delta_fkgl'], self.df['gemini_bert_f1'],
                    alpha=0.6, label='Gemini', color='blue')
        # Llama Data
        plt.scatter(self.df['llama_delta_fkgl'], self.df['llama_bert_f1'],
                    alpha=0.6, label='Llama', color='red')

        plt.title("Readability Improvement vs Semantic Preservation")
        plt.xlabel("Readability Improvement (Delta Flesch-Kincaid)")
        plt.ylabel("Semantic Fidelity (BERTScore F1)")
        plt.legend()
        plt.axhline(y=self.df['gemini_bert_f1'].mean(), color='blue', linestyle='--', alpha=0.3)
        plt.axhline(y=self.df['llama_bert_f1'].mean(), color='red', linestyle='--', alpha=0.3)
        plt.tight_layout()
        plt.show()

        # Box Plot Comparativo
        plt.figure(figsize=(10, 5))
        readability_data = pd.DataFrame({
            'Score': pd.concat([self.df['gemini_delta_fkgl'], self.df['llama_delta_fkgl']]),
            'Model': ['Gemini'] * len(self.df) + ['Llama'] * len(self.df)
        })
        sns.boxplot(x='Model', y='Score', data=readability_data, palette=['blue', 'red'])
        plt.title("Distribuzione Miglioramento Leggibilità (Delta FKGL)")
        plt.show()

        # Heatmap Correlazione
        plt.figure(figsize=(12, 5))

        plt.subplot(1, 2, 1)
        cols_gemini = ['gemini_delta_fkgl', 'gemini_bert_f1', 'gemini_cosine', 'gemini_fog']
        sns.heatmap(self.df[cols_gemini].corr(), annot=True, cmap='Blues', fmt=".2f")
        plt.title("Correlazione Metriche Gemini")

        plt.subplot(1, 2, 2)
        cols_llama = ['llama_delta_fkgl', 'llama_bert_f1', 'llama_cosine', 'llama_fog']
        sns.heatmap(self.df[cols_llama].corr(), annot=True, cmap='Reds', fmt=".2f")
        plt.title("Correlazione Metriche Llama")

        plt.tight_layout()
        plt.show()

        # Report Statistico
        print("\n=== REPORT STATISTICO ===")
        print(f"Gemini Mean BERTScore: {self.df['gemini_bert_f1'].mean():.4f}")
        print(f"Llama Mean BERTScore: {self.df['llama_bert_f1'].mean():.4f}")

        t_stat, p_val = stats.ttest_ind(self.df['gemini_bert_f1'], self.df['llama_bert_f1'])
        print(f"T-test (Semantic): p-value = {p_val:.4e} ({'Significativo' if p_val < 0.05 else 'Non significativo'})")

        # Analisi Outliers (Gemini)
        print("\n=== TOP OUTLIERS (High Readability, Low Semantic Fidelity) - Gemini ===")
        median_read = self.df['gemini_delta_fkgl'].median()
        p25_sem = self.df['gemini_bert_f1'].quantile(0.25)

        outliers = self.df[
            (self.df['gemini_delta_fkgl'] > median_read) &
            (self.df['gemini_bert_f1'] < p25_sem)
        ].head(5)

        if not outliers.empty:
            for idx, row in outliers.iterrows():
                print(f"\n--- ID: {row['text_id']} ---")
                print(f"Original: {row['original_text'][:100]}...")
                print(f"Simplified: {row['gemini_simplified'][:100]}...")
                print(f"Delta FKGL: {row['gemini_delta_fkgl']:.2f}, BERTScore: {row['gemini_bert_f1']:.4f}")
        else:
            print("Nessun outlier critico trovato con i criteri attuali.")

# ESECUZIONE
# Interfaccia input chiavi
print("\n Inserisci le tue chiavi API:")
try:
    # prende le chiavi dai Secrets di Colab
    GOOGLE_API_KEY = userdata.get('API_GOOGLE')
    HF_TOKEN = userdata.get('HF_TOKEN')
except:
    # Fallback input manuale
    GOOGLE_API_KEY = input("Inserisci Google API Key: ")
    HF_TOKEN = input("Inserisci HuggingFace Token: ")

if not GOOGLE_API_KEY or not HF_TOKEN:
    print(" Errore: Le chiavi API sono necessarie.")
else:
    # Istanzia ed esegui
    simplifier = LegalSimplifier(GOOGLE_API_KEY, HF_TOKEN)

    # Esegui esperimento
    df_results = simplifier.run_experiment()

    # Visualizza risultati
    simplifier.visualize_results()
```
